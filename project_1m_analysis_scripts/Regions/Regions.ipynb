```python
# Load the datasets from OceanDataCatalog

from OceanDataStore import OceanDataCatalog

catalog = OceanDataCatalog(catalog_name="noc-model-stac")

catalog.available_collections

#Era 5 probs the best here
```

```python
# Searching the ERA5 collection for sea surface temperature data

catalog.search(collection='noc-npd-era5', standard_name='sea_surface_temperature')

catalog.available_items

catalog.Items[5] 

# Near present run (1976 - 2024 (49 years))
# 0 for annual mean 1 degree
# 2 for annual mean 1/4 degree
# 5 for annual mean 1/12 degree
```

```python
# Create datasets (1, 025 and 12)

ds1 = catalog.open_dataset(id=catalog.Items[0].id,
                          start_datetime='1980-01',
                          end_datetime='1990-12',
                          )

ds1

ds025 = catalog.open_dataset(id=catalog.Items[2].id,
                          start_datetime='1980-01',
                          end_datetime='1990-12',
                          )

ds025

ds12 = catalog.open_dataset(id=catalog.Items[5].id,
                          start_datetime='1980-01',
                          end_datetime='1990-12',
                          )

ds12
```

```python
# ~36N (1-degree) 
import matplotlib.pyplot as plt

# Time mean
da_mean = ds1['tos_con'].mean(dim='time_counter')

vmin = -2.0
vmax = 30.0

x_start, x_end = 185, 296
y_start, y_end = 198, 312

y_kill_start = 241
y_kill_end = 287
x_kill_start = 208
x_kill_end = 281

DS_kill = 250

da_crop = da_mean.isel(
    x=slice(x_start, x_end),
    y=slice(y_start, y_end)
)

# Restore original coordinates on the cropped array
da_crop = da_crop.assign_coords(
    x=ds1['tos_con'].x[x_start:x_end],
    y=ds1['tos_con'].y[y_start:y_end]
)

# Plot
fig, ax = plt.subplots(figsize=(6, 6))
da_crop.plot(ax=ax, cmap='RdBu_r', vmin=vmin, vmax=vmax)
plt.title('Mean SST (time-averaged) – index slice')

y_line = da_crop['y'].isel(y=y_kill_start - y_start).item()
ax.axhline(y=y_line, color='black', linestyle='--', linewidth=2)

y_line = da_crop['y'].isel(y=y_kill_end - y_start).item()
ax.axhline(y=y_line, color='black', linestyle='--', linewidth=2)

x_line = da_crop['x'].isel(x=x_kill_start - x_start).item()
ax.axvline(x=x_line, color='black', linestyle='--', linewidth=2)

x_line = da_crop['x'].isel(x=x_kill_end - x_start).item()
ax.axvline(x=x_line, color='black', linestyle='--', linewidth=2)

x_line = da_crop['x'].isel(x=DS_kill - x_start).item()
ax.axvline(x=x_line, color='black', linestyle='--', linewidth=2)

plt.show()

# Convert to cropped array indices
y_idx = y_kill_start - y_start
x_idx_start = x_kill_start - x_start
x_idx_end = x_kill_end - x_start

# -----------------------------
# Slice the horizontal line segment (Dask-safe)
# -----------------------------
lat_line = da_crop['nav_lat'].isel(
    y=y_idx,
    x=slice(x_idx_start, x_idx_end)
).compute()  # forces Dask to load into memory

# -----------------------------
# Compute average latitude
# -----------------------------
avg_lat = lat_line.mean().item()  # now safe since it's a NumPy array
print("Average latitude:", avg_lat)



x_start, x_end = 270, 290
y_start, y_end = 230, 250

da_crop = da_mean.isel(
    x=slice(x_start, x_end),
    y=slice(y_start, y_end)
)

# Restore original coordinates on the cropped array
da_crop = da_crop.assign_coords(
    x=ds1['tos_con'].x[x_start:x_end],
    y=ds1['tos_con'].y[y_start:y_end]
)

# Plot
fig, ax = plt.subplots(figsize=(6, 6))
da_crop.plot(ax=ax, cmap='RdBu_r', vmin=vmin, vmax=vmax)
plt.title('Mean SST (time-averaged) – index slice')

y_line = da_crop['y'].isel(y=y_kill_start - y_start).item()
ax.axhline(y=y_line, color='black', linestyle='--', linewidth=2)

#y_line = da_crop['y'].isel(y=y_kill_end - y_start).item()
#ax.axhline(y=y_line, color='black', linestyle='--', linewidth=2)

#x_line = da_crop['x'].isel(x=x_kill_start - x_start).item()
#ax.axvline(x=x_line, color='black', linestyle='--', linewidth=2)

x_line = da_crop['x'].isel(x=x_kill_end - x_start).item()
ax.axvline(x=x_line, color='black', linestyle='--', linewidth=2)

plt.show()
```
